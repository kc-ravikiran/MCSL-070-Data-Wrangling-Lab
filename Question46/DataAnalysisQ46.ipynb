Question 46. Create Q45's graph and annotate the highest and lowest sales.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Plot order-wise sales.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(df["Order_ID"], df["Sales"], marker="o", label="Sales")

# Step 4: Label the plot and add a legend.
ax.set(xlabel="Order ID", ylabel="Sales (INR)", title="Sales by order")
ax.legend()


# Step 5: Locate the highest and lowest sales and annotate each point.
for label, row_index in [("Highest", df["Sales"].idxmax()), ("Lowest", df["Sales"].idxmin())]:
    row = df.loc[row_index]
    ax.annotate(
        f"{label}: {row['Order_ID']} ({row['Sales']:,})",
        xy=(row["Order_ID"], row["Sales"]), xytext=(15, -28 if label == "Highest" else 30),
        textcoords="offset points", arrowprops={"arrowstyle": "->"},
    )

# Step 6: Leave room around the plot for the annotation text.
ax.margins(x=0.16, y=0.18)

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question46_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question46_plot.png")
```
