Question 74. Plot the regression line between Income and Spending Score.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_PurchaseData.csv")
print("Original dataset:")
display(df)

# Step 3: Fit Spending Score as a straight-line function of Income.
fit = linregress(df["Income"], df["Spending Score"])

# Step 4: Plot observations with the fitted regression line.
fig, ax = plt.subplots(figsize=(9, 6))
sns.regplot(data=df, x="Income", y="Spending Score", ci=None, ax=ax)

# Step 5: Label the plot and report the line equation, correlation and R-squared.
ax.set(xlabel="Income (INR)", ylabel="Spending Score", title="Income versus Spending Score")
print(f"Spending Score = {fit.intercept:.4f} + ({fit.slope:.8f}) * Income")
print(f"Correlation: {fit.rvalue:.3f}; R-squared: {fit.rvalue ** 2:.3f}")

# Step 6: Save the fitted regression coefficients.
pd.DataFrame([{"Intercept": fit.intercept, "Slope": fit.slope, "R_squared": fit.rvalue ** 2}]).to_csv("regression_coefficients.csv", index=False)

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question74_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question74_plot.png")
```

Interpretation / method

The negative regression slope describes an association in this sample; it does not mean that higher income causes reduced spending.
