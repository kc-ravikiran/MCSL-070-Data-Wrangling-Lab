Question 70. Plot a histogram and univariate kernel density estimate (KDE) for Income.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_PurchaseData.csv")
print("Original dataset:")
display(df)

# Step 3: Plot an Income histogram on a density scale and overlay a smoothed KDE curve.
fig, ax = plt.subplots(figsize=(9, 5))
sns.histplot(data=df, x="Income", bins=5, stat="density", kde=True, ax=ax)

# Step 4: Add a descriptive title and labelled axes.
ax.set(xlabel="Income (INR)", ylabel="Density", title="Income histogram and KDE")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question70_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question70_plot.png")
```

Interpretation / method

The histogram and KDE share a density scale. Bandwidth and the ten-row sample strongly affect the KDE shape; no distribution is established.
