Question 73. Create a pairplot for the numeric variables and identify possible clusters or patterns.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
import seaborn as sns

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_PurchaseData.csv")
print("Original dataset:")
display(df)

# Step 3: Select numeric measurements, excluding the ID identifier.
numeric_columns = df.select_dtypes(include="number").columns.drop("ID").tolist()

# Step 4: Compare every numeric pair and colour points by purchase outcome.
grid = sns.pairplot(df, vars=numeric_columns, hue="Purchase", diag_kind="hist",
                   plot_kws={"s": 45, "alpha": 0.8}, height=1.8)

# Step 5: Add the plot title and display the numeric correlations.
fig = grid.fig
fig.suptitle("Numeric customer attributes by purchase outcome (ID excluded)", y=1.01)
display(df[numeric_columns].corr())

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question73_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question73_plot.png")
```

Interpretation / method

Purchasers tend to have higher Spending Score, Visits, Online Time and Rating in this sample; these variables are strongly related. ID is numeric but is not a meaningful measurement and is excluded. Visible groups in ten rows are not validated clusters.
