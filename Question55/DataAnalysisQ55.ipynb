Question 55. Use Latitude and Longitude to visualise the geographical distribution of sales.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Use longitude as x and latitude as y; bubble area and colour represent Sales.
fig, ax = plt.subplots(figsize=(11, 7))
points = ax.scatter(df["Longitude"], df["Latitude"], s=df["Sales"] / 25,
                    c=df["Sales"], cmap="viridis", alpha=0.8, edgecolors="black")

# Step 4: Label each bubble with its city name.
for row in df.itertuples():
    ax.annotate(row.City, (row.Longitude, row.Latitude), xytext=(5, 5), textcoords="offset points")

# Step 5: Add the sales colour scale and coordinate labels.
fig.colorbar(points, ax=ax, label="Sales (INR)")
ax.set(xlabel="Longitude (degrees east)", ylabel="Latitude (degrees north)",
       title="Sales by city (bubble area and colour represent sales)")

# Step 6: Adjust the display proportions using the mean latitude (this is not a projected map).
ax.set_aspect(1 / np.cos(np.deg2rad(df["Latitude"].mean())))

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question55_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question55_plot.png")
```

Interpretation / method

This is an offline coordinate plot, not a projected geographic basemap. Bangalore has the largest sales bubble.
