Question 59. Create a 2-by-2 grid: region-wise sales, category-wise profit, Age distribution, and Sales distribution.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create a 2-by-2 dashboard.
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# Step 4: Top left: total Sales by Region.
df.groupby("Region")["Sales"].sum().plot.bar(ax=axes[0, 0], color="steelblue", rot=0)
axes[0, 0].set(title="Region-wise sales", xlabel="Region", ylabel="Total sales (INR)")

# Step 5: Top right: total Profit by Category.
df.groupby("Category")["Profit"].sum().plot.bar(ax=axes[0, 1], color="seagreen", rot=0)
axes[0, 1].set(title="Category-wise profit", xlabel="Category", ylabel="Total profit (INR)")

# Step 6: Bottom left: distribution of customer ages.
axes[1, 0].hist(df["Age"], bins=5, edgecolor="black")
axes[1, 0].set(title="Age distribution", xlabel="Age (years)", ylabel="Customer count")

# Step 7: Bottom right: distribution of Sales.
axes[1, 1].hist(df["Sales"], bins=5, edgecolor="black")
axes[1, 1].set(title="Sales distribution", xlabel="Sales (INR)", ylabel="Order count")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question59_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question59_plot.png")
```
