Question 62. Plot Sales by Date and add a trendline.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Convert and sort the dates.
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m-%d")
ordered = df.sort_values("Date")

# Step 4: Plot the observed sales over time.
fig, ax = plt.subplots(figsize=(11, 5))

# Step 5: Overlay the trendline and report its slope in INR per day.
ax.plot(ordered["Date"], ordered["Sales"], marker="o", label="Sales")

# Step 6: Label the dates and sales values.
ax.set(xlabel="Date", ylabel="Sales (INR)", title="Sales over time: January 2026")
ax.legend()
fig.autofmt_xdate()


# Step 7: Represent each date as days since the first observation and fit a straight trendline.
days = (ordered["Date"] - ordered["Date"].min()).dt.days
slope, intercept = np.polyfit(days, ordered["Sales"], 1)
ax.plot(ordered["Date"], slope * days + intercept, "--", color="red", label="Linear trend")
ax.legend()
print(f"Trend slope: {slope:.2f} INR per day")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question62_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question62_plot.png")
```

Interpretation / method

The line is an ordinary least-squares descriptive trend for the ten observed orders, not a validated forecast.
