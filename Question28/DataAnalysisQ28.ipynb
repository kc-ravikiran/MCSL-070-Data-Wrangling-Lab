Question 28. Create overlapping fragments containing the first four and last four rows. Concatenate row-wise without duplicate rows.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Take the first four and last four rows; these fragments overlap.
first = df.head(4)
last = df.tail(4)

# Step 4: Combine the rows, remove duplicate records and reset row numbers.
combined = pd.concat([first, last], ignore_index=True)
result = combined.drop_duplicates().reset_index(drop=True)
display(first, last, result)

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```

Interpretation / method

The overlapping rows are employees 1003 and 1004. Removing duplicate complete rows restores the original six records.
