Question 58. Scatter Sales versus Profit using different colours and marker sizes for categories. Add a grid.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create a Sales/Profit chart.
fig, ax = plt.subplots(figsize=(9, 6))

# Step 4: Draw each category with its own colour and marker size.
for category, colour, size in zip(sorted(df["Category"].unique()),
                                  ["tab:blue", "tab:orange", "tab:green"], [60, 110, 170]):
    group = df.loc[df["Category"] == category]
    ax.scatter(group["Sales"], group["Profit"], c=colour, s=size, label=category, alpha=0.8)

# Step 5: Add axis labels, a light grid and a category legend.
ax.set(xlabel="Sales (INR)", ylabel="Profit (INR)", title="Sales and profit by category")
ax.grid(True, linestyle="--", alpha=0.4)
ax.legend(title="Category (colour and size)")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question58_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question58_plot.png")
```
