Question 40. Compute within-department rating Z-statistics and create ZRating.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Collect employee ratings within each department.
ratings = df.groupby("Department")["Rating"]

# Step 4: Find each employee's department mean and population standard deviation.
means = ratings.transform("mean")
stds = ratings.transform("std", ddof=0)

# Step 5: Calculate ZRating = (rating - department mean) / department standard deviation.
df["ZRating"] = (df["Rating"] - means) / stds

# Step 6: Display the new ZRating column.
result = df
display(result)

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```

Interpretation / method

ZRating=(Rating-department mean)/department population standard deviation (ddof=0). This convention gives each department mean 0 and population standard deviation 1. A zero-variance group would have undefined (NaN) Z-scores.
