Question 31. Swap Q30's index levels to Product at level 1 and Department at level 2. Display the result.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Recreate the Department/Product index from Q30.
indexed = df.set_index(["Department", "Product"]).sort_index()

# Step 4: Swap the two levels so Product comes before Department.
swapped = indexed.swaplevel("Department", "Product").sort_index()

# Step 5: Display the swapped index and prepare a flat export.
display(swapped)
result = swapped.reset_index()

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```
