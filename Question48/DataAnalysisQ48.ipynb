Question 48. Plot total Sales by Category with axis labels, title and legend. Annotate the highest bar.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Add the sales for each category and sort the totals.
totals = df.groupby("Category")["Sales"].sum().sort_values()

# Step 4: Draw a labelled bar chart with a legend.
fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(totals.index, totals.values, label="Total sales")
ax.set(xlabel="Category", ylabel="Total sales (INR)", title="Category sales totals")
ax.legend()

# Step 5: Find the tallest bar and label its total sales.
bar = bars[int(np.argmax(totals.values))]
ax.annotate(f"Highest: {totals.max():,}", (bar.get_x() + bar.get_width() / 2, bar.get_height()),
            xytext=(0, 10), textcoords="offset points", ha="center")

# Step 6: Leave room for the label and export category totals.
ax.margins(y=0.2)
totals.to_csv("category_sales.csv")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question48_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question48_plot.png")
```

Interpretation / method

Electronics has the highest total sales.
