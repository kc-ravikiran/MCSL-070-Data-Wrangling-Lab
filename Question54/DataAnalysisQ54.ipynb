Question 54. Convert Date to datetime and create a time-series line plot of sales.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Convert Date from text to datetime and sort observations chronologically.
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m-%d")
ordered = df.sort_values("Date")

# Step 4: Plot the observed Sales at their actual dates.
fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(ordered["Date"], ordered["Sales"], marker="o", label="Sales")

# Step 5: Label the time-series chart and format date labels.
ax.set(xlabel="Date", ylabel="Sales (INR)", title="Sales over time: January 2026")
ax.legend()
fig.autofmt_xdate()


# Step 6: Save the date-converted data without inventing missing daily records.
ordered.to_csv("dated_sales.csv", index=False)

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question54_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question54_plot.png")
```

Interpretation / method

The PDF provides ten orders on alternate days within January's first twenty days. Missing dates are not invented or assumed to have zero sales.
