Question 20. Correct employee 1003's salary to 33000. Check for salary outliers again.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_Data.csv")
print("Original dataset:")
display(df)

# Step 3: Correct only employee 1003's salary to 33000.
df.loc[df["Emp_ID"] == 1003, "Salary"] = 33000


# Step 4: Recalculate quartiles, IQR and outlier limits after the correction.
q1, q3 = df["Salary"].quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr

# Step 5: Find and display any remaining salary outliers.
outlier_mask = (df["Salary"] < lower) | (df["Salary"] > upper)
result = df.loc[outlier_mask]
print(f"Q1={q1:g}, Q3={q3:g}, IQR={iqr:g}; fences: {lower:g} to {upper:g}")
display(result)
print("Outlier employees:", result["Name"].tolist())


# Step 6: Save the corrected data separately; leave the input CSV unchanged.
df.to_csv("corrected_employee_Data.csv", index=False)
display(df)

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```

Interpretation / method

After correction, Q1=32000 and Q3=35000; the fences are 27500 and 39500. No salary is an outlier. The original source CSV remains unchanged.
