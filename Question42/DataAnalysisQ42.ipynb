Question 42. Create a Product-versus-Sales line graph. Discuss whether there is a trend.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Create a chart area and plot product names against sales.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(df["Product"], df["Sales"], marker="o")

# Step 4: Add labels and rotate product names so they remain readable.
ax.set(xlabel="Product", ylabel="Sales (INR)", title="Product sales")
ax.tick_params(axis="x", rotation=40)

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question42_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question42_plot.png")
```

Interpretation / method

Sales fluctuate across products; Laptop has the highest sales. Product names are nominal categories, so the connecting line does not establish a time trend.
