Question 51. Create a bar chart with headings showing product-category sales by region.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Total sales by Region and product Category; use zero for absent combinations.
totals = df.pivot_table(index="Region", columns="Category", values="Sales", aggfunc="sum", fill_value=0)

# Step 4: Plot category totals side-by-side within each region.
fig, ax = plt.subplots(figsize=(10, 5))
totals.plot.bar(ax=ax)

# Step 5: Add labels and a legend, then export the totals.
ax.set(xlabel="Region", ylabel="Total sales (INR)", title="Product-category sales by region")
ax.legend(title="Product category")
ax.tick_params(axis="x", rotation=0)
totals.to_csv("region_category_sales.csv")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question51_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question51_plot.png")
```

Interpretation / method

The table has Category rather than individual Product; categories are used as the available product grouping.
