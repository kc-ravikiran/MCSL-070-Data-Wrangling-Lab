Question 56. Represent Sales and Profit using box plots.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create separate box plots so different Sales and Profit scales stay readable.
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Step 4: Show each variable's median, quartiles and potential outliers.
for ax, column in zip(axes, ["Sales", "Profit"]):
    ax.boxplot(df[column])
    ax.set(xticks=[1], xticklabels=[column], ylabel="INR", title=f"{column} distribution")

# Step 5: Display numerical summaries alongside the plots.
display(df[["Sales", "Profit"]].describe())

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question56_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question56_plot.png")
```

Interpretation / method

Separate panels keep profit readable despite the larger sales scale. Box plots show medians, quartiles and any 1.5-IQR outliers.
