Question 52. Draw an appropriate graph showing Sales versus Profit.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Scatter Sales against Profit to compare the two numeric measurements.
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df["Sales"], df["Profit"])

# Step 4: Label the axes and report their correlation.
ax.set(xlabel="Sales (INR)", ylabel="Profit (INR)", title="Sales versus profit")
print(f"Correlation: {df['Sales'].corr(df['Profit']):.3f}")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question52_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question52_plot.png")
```

Interpretation / method

A scatter plot is appropriate for two quantitative variables. Higher sales are associated with higher profit in this sample.
