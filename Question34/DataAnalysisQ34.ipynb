Question 34. Group the employee performance data by Department and find mean Salary and Rating.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Group employees by Department and calculate mean Salary and Rating.
result = df.groupby("Department", as_index=False).agg(
    MeanSalary=("Salary", "mean"), MeanRating=("Rating", "mean")
)
display(result)

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```
