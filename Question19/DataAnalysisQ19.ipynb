Question 19. Find salary outliers using Q1, Q3 and the interquartile range (IQR). Name the employee whose salary is an outlier.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_Data.csv")
print("Original dataset:")
display(df)

# Step 3: Find the 25th percentile (Q1) and 75th percentile (Q3) of Salary.
q1, q3 = df["Salary"].quantile([0.25, 0.75])

# Step 4: Calculate IQR and the 1.5-IQR lower and upper outlier limits.
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr

# Step 5: Select salaries outside those limits and display the employee names.
outlier_mask = (df["Salary"] < lower) | (df["Salary"] > upper)
result = df.loc[outlier_mask]
print(f"Q1={q1:g}, Q3={q3:g}, IQR={iqr:g}; fences: {lower:g} to {upper:g}")
display(result)
print("Outlier employees:", result["Name"].tolist())

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```

Interpretation / method

Using Tukey's 1.5-IQR rule, Neha's original salary of 120000 is an outlier.
