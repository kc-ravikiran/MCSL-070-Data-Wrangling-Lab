Question 36. Group by Department and iterate over the groups, displaying each group's name and data.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Create one group for each department.
groups = df.groupby("Department")

# Step 4: Display and save each group's name and employee records.
for department, group in groups:
    print(f"Department: {department}")
    display(group)
    group.to_csv(f"{department}_employees.csv", index=False)

# Step 5: Summarise how many employees belong to each department.
result = groups.size().rename("Employees").reset_index()

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```
