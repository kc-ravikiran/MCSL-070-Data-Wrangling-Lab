Question 29. Create datasets (Emp_ID, Name, Department) and (Emp_ID, Department, Region, Sales). Concatenate column-wise on Emp_ID and explain how overlapping Department is handled.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create the first fragment and use Emp_ID to align rows.
first = df[["Emp_ID", "Name", "Department"]].set_index("Emp_ID")

# Step 4: Create the second fragment with the same employee index.
second = df[["Emp_ID", "Department", "Region", "Sales"]].set_index("Emp_ID")

# Step 5: Verify that Department agrees before keeping only one copy.
pd.testing.assert_series_equal(first["Department"], second["Department"])

# Step 6: Concatenate side-by-side, excluding the redundant Department column.
second_without_department = second.drop(columns="Department")
combined = pd.concat([first, second_without_department], axis=1)
result = combined.reset_index()
display(result)

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```

Interpretation / method

Emp_ID aligns rows. First verify that overlapping Department values agree, then keep one Department column rather than duplicate it.
