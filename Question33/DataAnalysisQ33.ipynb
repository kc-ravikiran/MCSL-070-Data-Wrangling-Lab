Question 33. Create the Department/Product multilevel index, stack columns into rows, display the result, and revert to the earlier format.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create the two-level Department/Product index.
indexed = df.set_index(["Department", "Product"])

# Step 4: Add Emp_ID temporarily because several employees share Department/Product.
# Unique row labels let us unstack later without combining different employees.
unique_indexed = indexed.set_index("Emp_ID", append=True)
stacked = unique_indexed.stack(future_stack=True)
display(stacked)
stacked.rename("Value").to_csv("stacked_data.csv")

# Step 5: Unstack without sorting, then move Emp_ID back into a normal column.
restored = stacked.unstack(sort=False).reset_index("Emp_ID")
restored = restored.reindex(columns=indexed.columns).astype(indexed.dtypes)

# Step 6: Display the reconstructed table and prepare a flat CSV export.
display(restored)
result = restored.reset_index()

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```

Interpretation / method

Some Department/Product pairs repeat, so Emp_ID is temporarily added as a third index level to keep each employee unique during stacking. Unstack then restores Emp_ID as a column and the original two-level index. Original dtypes are restored because stacking mixed text and numbers produces object dtype.
