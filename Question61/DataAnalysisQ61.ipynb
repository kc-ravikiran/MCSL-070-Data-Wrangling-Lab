Question 61. Create a customised plot setting, including font size and plot size.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Apply temporary font, figure-size and grid settings, then plot regional sales.
with plt.rc_context({"font.size": 12, "axes.titlesize": 16, "axes.labelsize": 13,
                     "figure.figsize": (11, 6), "axes.grid": True, "grid.alpha": 0.25}):
    fig, ax = plt.subplots()
    totals = df.groupby("Region")["Sales"].sum().sort_values()
    ax.bar(totals.index, totals.values, color="teal", edgecolor="black")
    ax.set(xlabel="Region", ylabel="Total sales (INR)", title="Customised regional sales chart")
    ax.tick_params(axis="x", rotation=0)

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question61_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question61_plot.png")
```

Interpretation / method

rc_context scopes the custom style so other notebooks are unaffected.
