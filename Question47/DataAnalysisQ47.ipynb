Question 47. Create Q46's graph. Save high-resolution PNG and JPG copies.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Plot order-wise sales.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(df["Order_ID"], df["Sales"], marker="o", label="Sales")

# Step 4: Add a title, labels and legend.
ax.set(xlabel="Order ID", ylabel="Sales (INR)", title="Sales by order")
ax.legend()


# Step 5: Annotate the maximum and minimum order sales.
for label, row_index in [("Highest", df["Sales"].idxmax()), ("Lowest", df["Sales"].idxmin())]:
    row = df.loc[row_index]
    ax.annotate(
        f"{label}: {row['Order_ID']} ({row['Sales']:,})",
        xy=(row["Order_ID"], row["Sales"]), xytext=(15, -28 if label == "Highest" else 30),
        textcoords="offset points", arrowprops={"arrowstyle": "->"},
    )

# Step 6: Leave space for labels and save the same figure as 300-DPI PNG and JPG files.
ax.margins(x=0.16, y=0.18)

fig.tight_layout()
fig.savefig("sales_high_resolution.png", dpi=300, bbox_inches="tight")
fig.savefig("sales_high_resolution.jpg", dpi=300, bbox_inches="tight")
print("Saved PNG and JPG images at 300 DPI.")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question47_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question47_plot.png")
```

Interpretation / method

Both files are generated directly from the same figure at 300 DPI; the JPG is not a recompressed low-resolution screenshot.
