Question 26. Split employee_SalesData.csv into data1set (Emp_ID, Name, Department, Year) and data2set (Emp_ID, Region, Product, Sales). Remove employee 1004 from data1set and 1002 from data2set. Perform an inner join and verify missing records.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create the first fragment and exclude employee 1004.
data1set = df.loc[df["Emp_ID"] != 1004, ["Emp_ID", "Name", "Department", "Year"]].copy()

# Step 4: Create the second fragment and exclude employee 1002.
data2set = df.loc[df["Emp_ID"] != 1002, ["Emp_ID", "Region", "Product", "Sales"]].copy()

# Step 5: Save and display both fragments.
data1set.to_csv("data1set.csv", index=False)
data2set.to_csv("data2set.csv", index=False)

display(data1set, data2set)

# Step 6: Inner-join on Emp_ID: retain only employees present in both fragments.
result = data1set.merge(data2set, on="Emp_ID", how="inner", validate="one_to_one")
display(result)

# Step 7: Compare employee IDs to identify which original records are missing.
print("Missing employee IDs:", sorted(set(df["Emp_ID"]) - set(result["Emp_ID"])))

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```

Interpretation / method

The inner join keeps IDs present in both fragments: 1001, 1003, 1005 and 1006. IDs 1002 and 1004 are missing.
