Question 50. Create a Sales-versus-Rating scatter plot with a regression line. Describe the trend.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Fit a straight line relating Sales to Rating.
slope, intercept = np.polyfit(df["Sales"], df["Rating"], 1)

# Step 4: Create sales values spanning the observed range for the fitted line.
x = np.linspace(df["Sales"].min(), df["Sales"].max(), 100)

# Step 5: Plot observed ratings and the fitted regression line.
fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(df["Sales"], df["Rating"], label="Orders")
ax.plot(x, slope * x + intercept, color="red", label="Linear fit")

# Step 6: Add chart labels and report the correlation and line slope.
ax.set(xlabel="Sales (INR)", ylabel="Rating", title="Sales versus rating")
ax.legend()
print(f"Correlation: {df['Sales'].corr(df['Rating']):.3f}; slope: {slope:.8f}")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question50_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question50_plot.png")
```

Interpretation / method

The fitted line shows a modest positive association, but ratings vary even among low-sales orders. Correlation is descriptive, not evidence that increasing sales causes better ratings.
