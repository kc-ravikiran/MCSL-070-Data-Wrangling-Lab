Question 43. Create two subplots: Price versus Sales, and Discount versus Sales. Discuss similarities.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Create two chart areas side-by-side.
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Step 4: Compare Sales with Price and Discount, using a labelled scatter plot for each.
for ax, column in zip(axes, ["Price", "Discount (%)"]):
    ax.scatter(df[column], df["Sales"])
    ax.set(xlabel=f"{column} (INR)" if column == "Price" else column, ylabel="Sales (INR)", title=f"{column} versus Sales")
    print(f"{column}/Sales correlation: {df[column].corr(df['Sales']):.3f}")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question43_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question43_plot.png")
```

Interpretation / method

Price has a strong positive association with Sales; Discount does not show the same pattern. Sales also depends on quantity and discounted price, so these plots do not isolate a causal discount effect.
