Question 45. Plot Order_ID versus Sales with axis labels, title and legend.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Plot Sales for each Order_ID using a line with markers.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(df["Order_ID"], df["Sales"], marker="o", label="Sales")

# Step 4: Add a title, axis labels and a legend.
ax.set(xlabel="Order ID", ylabel="Sales (INR)", title="Sales by order")
ax.legend()

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question45_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question45_plot.png")
```
