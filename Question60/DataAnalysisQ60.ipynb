Question 60. Annotate minimum and maximum Sales and minimum and maximum Profit in the related graphs.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create one chart for Sales and another for Profit.
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Step 4: Plot order values and label each chart's minimum and maximum.
for ax, column in zip(axes, ["Sales", "Profit"]):
    ax.bar(df["OID"], df[column], color="steelblue")
    ax.set(xlabel="Order ID", ylabel=f"{column} (INR)", title=f"Minimum and maximum {column.lower()}")
    for label, idx in [("Min", df[column].idxmin()), ("Max", df[column].idxmax())]:
        row = df.loc[idx]
        ax.annotate(f"{label}: {row[column]:,}", (row["OID"], row[column]),
                    xytext=(0, 20), textcoords="offset points", ha="center",
                    arrowprops={"arrowstyle": "->"})
    ax.margins(y=0.25)

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question60_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question60_plot.png")
```
