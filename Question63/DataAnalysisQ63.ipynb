Question 63. Create a 3D plot of Sales, Profit and Age. Change the variables on the axes to visualise different views.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create a figure for two 3D views.
fig = plt.figure(figsize=(14, 6))

# Step 4: Plot Sales, Profit and Age with different axis assignments and viewing angles.
for position, columns in enumerate([("Sales", "Profit", "Age"), ("Age", "Sales", "Profit")], start=1):
    ax = fig.add_subplot(1, 2, position, projection="3d")
    x, y, z = columns
    ax.scatter(df[x], df[y], df[z], c=df["Age"], cmap="viridis", s=65)
    ax.set(xlabel=x, ylabel=y, zlabel=z, title=f"{x}, {y}, {z}")
    ax.view_init(elev=25, azim=35 if position == 1 else 125)

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question63_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question63_plot.png")
```

Interpretation / method

Both axis assignment and viewing angle are changed to inspect the same observations from different perspectives.
