Question 53. Create a scatter plot matrix for Sales, Profit and Age. Discuss possible relationships.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
import seaborn as sns

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create a scatter plot matrix for Sales, Profit and Age with histograms on the diagonal.
grid = sns.pairplot(df[["Sales", "Profit", "Age"]], diag_kind="hist")

# Step 4: Add a title and display correlations to help interpret the relationships.
fig = grid.fig
fig.suptitle("Sales, profit and age", y=1.02)
display(df[["Sales", "Profit", "Age"]].corr())

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question53_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question53_plot.png")
```

Interpretation / method

Sales and Profit have a strong positive association. The age relationships are weaker; this ten-row sample does not support a reliable age-based conclusion.
