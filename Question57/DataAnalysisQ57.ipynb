Question 57. Create side-by-side box plots of Profit by Category and compare the categories.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("Customer_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: List categories in a consistent order.
categories = sorted(df["Category"].unique())

# Step 4: Collect each category's profits and draw side-by-side box plots.
fig, ax = plt.subplots(figsize=(9, 5))
ax.boxplot([df.loc[df["Category"] == category, "Profit"] for category in categories])

# Step 5: Label the boxes and display category profit summaries.
ax.set(xticks=range(1, len(categories) + 1), xticklabels=categories, xlabel="Category",
       ylabel="Profit (INR)", title="Profit by category")
display(df.groupby("Category")["Profit"].agg(["count", "median", "min", "max"]))

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question57_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question57_plot.png")
```

Interpretation / method

Electronics has the highest median profit, followed by Clothing and Grocery. Each category contains only three or four observations.
