Question 39. Group by Department and calculate mean, total and maximum Salary and Sales.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Calculate mean, sum and maximum Salary and Sales within each department.
summary = df.groupby("Department")[["Salary", "Sales"]].agg(["mean", "sum", "max"])
display(summary)

# Step 4: Give the summary columns simple names for CSV export.
column_names = []
for column, statistic in summary.columns:
    column_names.append(f"{column}_{statistic}")
summary.columns = column_names
result = summary.reset_index()

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```
