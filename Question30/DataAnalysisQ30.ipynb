Question 30. Create a two-level hierarchical index: Department at level 1 and Product at level 2. Display the indexed data.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Create and sort the Department/Product hierarchical row index.
indexed = df.set_index(["Department", "Product"]).sort_index()

# Step 4: Display the indexed data and prepare a flat copy for CSV export.
display(indexed)
result = indexed.reset_index()

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```
