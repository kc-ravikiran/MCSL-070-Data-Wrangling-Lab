Question 49. Create histograms for Sales and Price. Analyse both distributions.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Create separate panels for the Sales and Price distributions.
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Step 4: Draw histograms showing how many orders fall into each value range.
for ax, column in zip(axes, ["Sales", "Price"]):
    ax.hist(df[column], bins=5, edgecolor="black")
    ax.set(xlabel=f"{column} (INR)", ylabel="Order count", title=f"{column} distribution")

# Step 5: Compare mean, median and skewness to describe both distributions.
print(df[["Sales", "Price"]].agg(["mean", "median", "skew"]))

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question49_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question49_plot.png")
```

Interpretation / method

Both distributions are right-skewed: most orders have relatively low values, with the Laptop order producing a long upper tail. Ten observations are too few to establish a population distribution.
