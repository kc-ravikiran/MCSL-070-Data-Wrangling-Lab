Question 44. Plot Price versus Sales using a dotted green line and triangle markers.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_DataPerformance.csv")
print("Original dataset:")
display(df)

# Step 3: Sort by Price so the connecting line follows increasing prices.
ordered = df.sort_values("Price")

# Step 4: Draw the requested green dotted line with triangle markers.
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(ordered["Price"], ordered["Sales"], color="green", linestyle=":", marker="^")

# Step 5: Add a clear title and axis labels.
ax.set(xlabel="Price (INR)", ylabel="Sales (INR)", title="Price versus sales")

# Final step: Arrange labels, save the chart and display it below this cell.
fig.tight_layout()
fig.savefig("question44_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question44_plot.png")
```
