Question 32. Index by Emp_ID and pivot with Product as columns and Sales as values. Restore the original format.

```python
# Step 1: Import the libraries needed for this program.
import pandas as pd
from IPython.display import display

# Step 2: Load the CSV file from this question's folder into a DataFrame (table).
df = pd.read_csv("employee_SalesData.csv")
print("Original dataset:")
display(df)

# Step 3: Use employee IDs as row labels and product names as pivot columns.
indexed = df.set_index("Emp_ID")
wide = indexed.pivot(columns="Product", values="Sales")

# Step 4: Display and save the wide-format sales table.
display(wide)
wide.to_csv("pivot_sales.csv")

# Step 5: Melt product columns back into Product and Sales rows, omitting empty combinations.
long = wide.reset_index().melt(id_vars="Emp_ID", var_name="Product", value_name="Sales")
long = long.dropna(subset=["Sales"])
long["Sales"] = long["Sales"].astype(df["Sales"].dtype)

# Step 6: Restore employee details retained separately from the pivot.
metadata = df.drop(columns=["Product", "Sales"])
result = metadata.merge(long, on="Emp_ID", validate="one_to_one")[df.columns]

# Step 7: Restore the original employee order and display the reconstructed dataset.
result = result.sort_values("Emp_ID").reset_index(drop=True)
display(result)

# Final step: Save the answer without the automatic row-number column.
result.to_csv("result.csv", index=False)
print("Saved result.csv")
```

Interpretation / method

Non-sales metadata is retained separately so the full original DataFrame, not only the three pivoted fields, can be restored.
