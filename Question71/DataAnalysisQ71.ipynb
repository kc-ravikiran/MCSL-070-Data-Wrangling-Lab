Question 71. Plot a bivariate KDE for Income against Spending Score.

```python
%matplotlib inline
import matplotlib.pyplot as plt
from pathlib import Path
import os
import pandas as pd
import numpy as np
from IPython.display import display

DATA_DIR = Path.cwd() if Path("Customer_PurchaseData.csv").exists() else Path("Question71")
df = pd.read_csv(DATA_DIR / "Customer_PurchaseData.csv")
os.chdir(DATA_DIR)
display(df)

import seaborn as sns
fig, ax = plt.subplots(figsize=(9, 6))
sns.kdeplot(data=df, x="Income", y="Spending Score", fill=True, levels=8,
            thresh=0.05, cmap="Blues", ax=ax, cut=0)
ax.scatter(df["Income"], df["Spending Score"], c="black", s=30, label="Observed customers")
ax.set(xlabel="Income (INR)", ylabel="Spending Score", title="Income and Spending Score: bivariate KDE")
ax.legend()

fig.tight_layout()
fig.savefig("question71_plot.png", dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)
print("Saved question71_plot.png")
```

Interpretation / method

The observations are overlaid so smoothing is not mistaken for additional data. Higher incomes generally have lower spending scores in this sample.
